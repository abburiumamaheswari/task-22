```python
import pandas as pd
```

```python
data = {
    "Department": ["IT", "HR", "Finance", "IT", "HR", "Finance", "IT", "HR"],
    "Employee": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Salary": [50000, 40000, 60000, 55000, 45000, 65000, 70000, 48000],
    "Experience": [2, 3, 5, 4, 2, 6, 7, 3]
}

df = pd.DataFrame(data)

print(df)
```

```python
print("Sum of Salary by Department")

group_sum = df.groupby("Department")["Salary"].sum()

print(group_sum)
```

```python
print("Average Salary by Department")

group_mean = df.groupby("Department")["Salary"].mean()

print(group_mean)
```

```python
print("Employee Count in Each Department")

group_count = df.groupby("Department")["Employee"].count()

print(group_count)
```

```python
print("Minimum Salary in Each Department")

group_min = df.groupby("Department")["Salary"].min()

print(group_min)
```

```python
print("Maximum Salary in Each Department")

group_max = df.groupby("Department")["Salary"].max()

print(group_max)
```

```python
print("Department and Experience")

group_two = df.groupby(["Department", "Experience"])["Salary"].sum()

print(group_two)
```

```python
print("Multiple Aggregations")

result = df.groupby("Department")["Salary"].agg(["sum", "mean", "min", "max", "count"])

print(result)
```
